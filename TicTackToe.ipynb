```python
import pygame
import random
import sys

# -----------------------------
# Settings
# -----------------------------

WIDTH = 600
HEIGHT = 700
CELL_SIZE = 200
FPS = 60

pygame.init()

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Tic-Tac-Toe")

clock = pygame.time.Clock()

font = pygame.font.Font(None, 120)
message_font = pygame.font.Font(None, 50)

# -----------------------------
# Game variables
# -----------------------------

board = [""] * 9

player = "X"
computer = "O"

current_turn = player

game_over = False
winner = None


# -----------------------------
# Draw board
# -----------------------------

def draw_board():

    screen.fill("white")

    # Grid
    for i in range(1, 3):

        pygame.draw.line(
            screen, "black",
            (i * CELL_SIZE, 0),
            (i * CELL_SIZE, 600),
            5
        )

        pygame.draw.line(
            screen, "black",
            (0, i * CELL_SIZE),
            (600, i * CELL_SIZE),
            5
        )

    # X and O
    for i, value in enumerate(board):

        if value != "":

            row = i // 3
            col = i % 3

            x = col * CELL_SIZE + CELL_SIZE // 2
            y = row * CELL_SIZE + CELL_SIZE // 2

            text = font.render(value, True, "black")

            screen.blit(
                text,
                text.get_rect(center=(x, y))
            )


# -----------------------------
# Check winner
# -----------------------------

def check_winner():

    combinations = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in combinations:

        if board[a] == board[b] == board[c] != "":
            return board[a]

    if "" not in board:
        return "Draw"

    return None


# -----------------------------
# Computer move
# -----------------------------

def computer_move():

    empty = [
        i for i in range(9)
        if board[i] == ""
    ]

    if empty:

        position = random.choice(empty)

        board[position] = computer


# -----------------------------
# Restart
# -----------------------------

def restart():

    global board
    global current_turn
    global game_over
    global winner

    board = [""] * 9

    current_turn = player

    game_over = False
    winner = None


# -----------------------------
# Main game loop
# -----------------------------

running = True

while running:

    # -------------------------
    # Events
    # -------------------------

    for event in pygame.event.get():

        if event.type == pygame.QUIT:
            running = False

        # Keyboard
        if event.type == pygame.KEYDOWN:

            if event.key == pygame.K_q:
                running = False

            if event.key == pygame.K_r:
                restart()

        # Mouse
        if event.type == pygame.MOUSEBUTTONDOWN:

            if not game_over and current_turn == player:

                x, y = event.pos

                # Click inside board
                if y < 600:

                    col = x // CELL_SIZE
                    row = y // CELL_SIZE

                    position = row * 3 + col

                    # Empty square?
                    if board[position] == "":

                        board[position] = player

                        winner = check_winner()

                        if winner:

                            game_over = True

                        else:

                            current_turn = computer


    # -------------------------
    # Computer
    # -------------------------

    if not game_over and current_turn == computer:

        computer_move()

        winner = check_winner()

        if winner:

            game_over = True

        else:

            current_turn = player


    # -------------------------
    # Draw
    # -------------------------

    draw_board()

    # Message
    if game_over:

        if winner == "Draw":
            message = "Draw! Press R"

        elif winner == player:
            message = "You Win! Press R"

        else:
            message = "Computer Wins! Press R"

    else:

        if current_turn == player:
            message = "Your Turn"

        else:
            message = "Computer Turn"

    text = message_font.render(
        message,
        True,
        "black"
    )

    screen.blit(
        text,
        text.get_rect(center=(300, 650))
    )

    pygame.display.flip()

    clock.tick(FPS)


pygame.quit()
sys.exit()
```
